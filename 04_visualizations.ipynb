# Phase 4: Data Visualizations & Insights
- **Objective:** Build univariate, bivariate, and multivariate visualizations using Matplotlib, Seaborn, and Plotly, ensuring each chart is accompanied by a professional business insight[cite: 1].

```python
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Set style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load data from SQLite database
db_path = '/content/data/tmdb_movies.db'
conn = sqlite3.connect(db_path)
movies_df = pd.read_sql("SELECT * FROM movies", conn)
conn.close()

print("Libraries loaded and data fetched for visualizations!")
```

```python
# --- Chart 1: Distribution of Movie Budgets ---
plt.figure()
sns.histplot(movies_df[movies_df['budget'] > 0]['budget'] / 1e6, bins=30, kde=True, color='skyblue')
plt.title('Distribution of Movie Budgets (in Millions USD)')
plt.xlabel('Budget ($ Millions)')
plt.ylabel('Count')
plt.show()
```

### Chart 1 Insight:
- Movie production budgets are heavily right-skewed. The vast majority of films operate on modest financial scales, while rare high-risk blockbuster outliers exceed $200M[cite: 1].

```python
# --- Chart 2: Distribution of Runtimes ---
plt.figure()
sns.boxplot(x=movies_df['runtime'], color='lightgreen')
plt.title('Distribution of Movie Runtimes')
plt.xlabel('Runtime (Minutes)')
plt.show()
```

### Chart 2 Insight:
- Standard feature films cluster strictly within the 90 to 120-minute window to optimize theatrical screening rotations, with rare epic runtimes acting as statistical outliers[cite: 1].

```python
# --- Chart 3: Budget vs Revenue Scatter Plot ---
valid_df = movies_df[(movies_df['budget'] > 0) & (movies_df['revenue'] > 0)].copy()
plt.figure()
sns.scatterplot(x=valid_df['budget']/1e6, y=valid_df['revenue']/1e6, alpha=0.6, color='purple')
plt.title('Budget vs Box-Office Revenue')
plt.xlabel('Budget ($ Millions)')
plt.ylabel('Revenue ($ Millions)')
plt.show()
```

### Chart 3 Insight:
- A positive correlation exists between higher production spend and box-office returns, though high investment does not guarantee profitability, as several high-budget films underperform[cite: 1].

```python
# --- Chart 4: Yearly Movie Output ---
movies_df['release_year'] = pd.to_datetime(movies_df['release_date'], errors='coerce').dt.year
yearly_counts = movies_df['release_year'].value_counts().sort_index()

plt.figure()
yearly_counts.tail(30).plot(kind='bar', color='salmon')
plt.title('Number of Movies Released Per Year')
plt.xlabel('Release Year')
plt.ylabel('Movie Count')
plt.xticks(rotation=45)
plt.show()
```

### Chart 4 Insight:
- Global film production output expanded rapidly over recent decades, peaking around the mid-2010s due to digital filmmaking accessibility before stabilizing[cite: 1].
